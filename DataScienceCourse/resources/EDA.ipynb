Exploratory Data Analysis (EDA) is an important step in data analysis where we explore, summarize, and visualize data to understand its structure, detect patterns, identify anomalies, test assumptions, and check relationships between variables before applying any machine learning or statistical models

# Types of Exploratory Data Analysis

#1. Univariate Analysis

-> Univariate analysis studies one variable at a time to understand its characteristics and distribution.

Histograms

Box Plots

Bar charts

#2. Bivariate Analysis
-> Bivariate analysis examines the relationship between two variables to understand how they interact or influence each other.

Scatter plots

Correlation coefficient

Cross-tabulation

Line graphs

Covariance

#3. Multivariate Analysis
->Multivariate analysis studies three or more variables together to understand complex relationships within the dataset.

Pair Plots

Principal Component Analysis

Spatial Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the built-in iris dataset
iris = sns.load_dataset("iris")

# Set up the figure size for a 2x2 grid
plt.figure(figsize=(12, 10))

# ----------------------------------------------------
# Position 1: Top-Left (Row 1, Column 1)
# ----------------------------------------------------
plt.subplot(3, 2, 1)
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("1. Sepal Length vs Width")

# ----------------------------------------------------
# Position 2: Top-Right (Row 1, Column 2) -> Your target position
# ----------------------------------------------------
plt.subplot(3, 2, 2)
sns.boxplot(data=iris, x="species", y="petal_length", hue="species", legend=False)
plt.title("2. Petal Length by Species")

# ----------------------------------------------------
# Position 3: Bottom-Left (Row 2, Column 1)
# ----------------------------------------------------
plt.subplot(3, 2, 3)
sns.histplot(data=iris, x="petal_width", hue="species", kde=True, multiple="stack")
plt.title("3. Petal Width Distribution")

# ----------------------------------------------------
# Position 4: Bottom-Right (Row 2, Column 2)
# ----------------------------------------------------
plt.subplot(3, 2, 4)
sns.violinplot(data=iris, x="species", y="sepal_length", hue="species", legend=False)
plt.title("4. Sepal Length Violin Plot")

# Automatically adjust spacing between plots so labels don't overlap
plt.tight_layout()

# Display the complete grid
plt.show()
```

#Detecting and Removing Outliers

1. Method 1: IQR(Interquartile Range Rule)

```python
Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Valid range: {lower_bound} to {upper_bound}")
print()
outliers = df[(df["income"] < lower_bound) | (df["income"] > upper_bound)]
print(outliers)
```

Removing IQR based outliers

```python
df_clean = df[(df["income"] >= lower_bound) & (df["income"] <= upper_bound)]
print(f"Original rows: {len(df)}, After removing outliers: {len(df_clean)}")
```

2. Method 2-> Z-Score Rule

```python
from scipy import stats

df["income_zscore"] = stats.zscore(df["income"])
print(df[["income", "income_zscore"]])

outliers_z = df[df["income_zscore"].abs() > 3]
print(outliers_z)

df_clean_z = df[df["income_zscore"].abs() <= 3]
```

Z-Score
 measures how many standard deviations a value is from the mean; a common rule of thumb flags anything beyond 3 standard deviations as an outlier

```python
# Capping instead of removing — an alternative worth showing
df["income_capped"] = df["income"].clip(lower=lower_bound, upper=upper_bound)
print(df[["income", "income_capped"]])
```
