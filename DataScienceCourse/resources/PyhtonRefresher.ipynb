Functions

```python
#Functions
def greet(name):
    print(f"Hello {name}")

#call
greet("ram")
```

```python
#Default Arguments
def greet(name, city="Pokhara"):
    print(f"{name} lives in {city}")

greet("Ram")
greet("Sita", "Kathmandu")
```

```python
#Keyword Arguments
def student(name, age):
    print(name, age)

student(age=20, name="Hari")
```

List Comprehension

```python
#List Comprehension
numbers = [1, 2, 3, 4, 5]

squares = []

for n in numbers:
    squares.append(n * n)

print(squares)
```

```python
#Pythonic way
squares = [n * n for n in numbers]

print(squares)
```

Filtering


```python
#Traditional approach
even = []

for n in numbers:
    if n % 2 == 0:
        even.append(n)
print(even)
```

```python
# pythonic way
even = [n for n in numbers if n % 2 == 0]
print(even)
```

Dictionary Comprehension


```python
numbers = [1, 2, 3]

square_dict = {n: n*n for n in numbers}

print(square_dict)
```

Set Comprehension (distinct values , no duplicates)

```python
numbers1 = [1,2,2,3,3,4]

unique = {n for n in numbers1}

print(unique)
```

*args and **kwargs

```python
# Sometimes we don't know how many inputs a function will receive
#*args
def total(*numbers):
    print(sum(numbers))

total(10,20)
total(10,20,30,40)
```

*args collects positional arguments into a tuple.

```python
#**kwargs
def student(**info):
    print(info)

student(name="Ram", age=20, city="Pokhara")
```

**kwargs collects keyword arguments into a dictionary.
