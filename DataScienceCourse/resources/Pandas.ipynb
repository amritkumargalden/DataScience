Series from a dictionary

```python
import pandas as pd

# create a dictionary
grades = {"Semester1": 3.25, "Semester2": 3.28, "Semester3": 3.75}

# create a series from the dictionary
my_series = pd.Series(grades,index=('Semester1','Semester2','Semester3'))

# display the series
print(my_series)
```

```python

# Custom index puts Semester3 first
series_reordered = pd.Series(grades, index=['Semester3', 'Semester1', 'Semester2'])
print(series_reordered)
```

```python
# Leaving out 'Semester2' filters it out entirely
series_filtered = pd.Series(grades, index=['Semester1', 'Semester3'])
print(series_filtered)
```

```python
import pandas as pd
import numpy as np

# Series A has indices: a, b, c (with a missing value in b)
series_a = pd.Series([10, np.nan, 30], index=['a', 'b', 'c'])

# Series B has indices: b, c, d (missing index 'a', introduces index 'd')
series_b = pd.Series([2, 3, 4], index=['b', 'c', 'd'])

# Standard multiplication results in NaN for missing indices or values
standard_mult = series_a * series_b

# Using .mul() with fill_value=1 replaces missing data before multiplying
filled_mult = series_a.mul(series_b, fill_value=1)

print("--- Standard Multiplication (*) ---")
print(standard_mult)

print("\n--- Multiplication with fill_value=1 ---")
print(filled_mult)
```

## header
-> Tells pandas which row contains your column names(headers). By default, header=0(the very first row)

Sometimes, a CSV file includes useless introductory text (like "School Data Report") at the very top before the real columns start. You must skip these and explicitly tell Pandas where the headers are.

Example: If the actual column names are on the third row(index = 2)

df=pd.read_csv("file_name",header=2

# na_values
-> Tells Pandas which strings in your dataset represent missing data so it can convert them to standard "Not a Number"(NaN) values for calclualtions

Real-world files are messy. Instead of blank spaces, missing grades might be marked with an N/A, a hyphen (-), or a question mark (?). If not handled, Pandas will read those columns as text strings rather than numbers, breaking your math

Example:
missing_values = ["N/A", "-", "?"]

df = pd.read_csv("student_data.csv", na_values=missing_values)
