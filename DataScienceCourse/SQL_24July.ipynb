# GroupBy

```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "city": ["Kathmandu", "Pokhara", "Kathmandu", "Biratnagar", "Pokhara", "Kathmandu"],
    "category": ["Electronics", "Grocery", "Grocery", "Electronics", "Electronics", "Grocery"],
    "amount": [4500, 1200, 800, 6200, 3100, 950]
})
print(orders)
```

```python
grouped = orders.groupby("city")
print(grouped)
```

```python
print(grouped["amount"].mean())
```

```python
print(grouped["amount"].sum())
```

```python
## Grouping by multiple columns
```

```python
multi_group = orders.groupby(["city", "category"])["amount"].mean(
)
print(multi_group)
```

```python
print(orders.groupby("city")
      ["amount"].min()
      )
print()
print(orders.groupby("city")
      ["amount"].max()
      )
print()
print(orders.groupby("city")
      ["amount"].std()
      )
print()
```

```python
# By default, groupby results in indexing the column as index
# to reset Index , use reset_index()

result = orders.groupby("city")["amount"].mean().reset_index()
print(result)
print(type(result))
```

# Aggregate functions

```python
# .agg()

print(orders.groupby("city")["amount"].agg(["sum", "mean","count"])
      )
```

```python
summary = orders.groupby("city").agg(
    {
        "amount": ["sum","mean"],
        "order_id": "count"
    }
)

print(summary)
```

```python
summary_named = orders.groupby("city").agg(
    {
        total_revenue = ("amount",)
    }
)

print(summary_named)
```

# Pivot Tables

```python
pivot = orders.pivot_table(
    values = "amount",
    index="city",
    columns="category",
    aggfunc = "sum",
    fill_value=0
)

print(pivot)
```
