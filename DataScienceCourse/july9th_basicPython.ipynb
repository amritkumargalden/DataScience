```python
numbers = [1,2,3,4,5]
even_numbers = []
```

```python
for n in numbers: 
    if n % 2 == 0:
        print(f"{n} is even")
        n = n+2
        even_numbers.append(n)
    else:
        print(f"{n} is odd")

print(even_numbers)
```

```python
even = [n for n in numbers if n % 2 == 0 ]
print(even)
```

```python
# *args and **kwargs

def print_args(*args):
    for arg in args:
        print(arg)

print_args("Hello", "World", "Python")

#**kwargs
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_kwargs(name="Amrit", age=25, role="Data Engineer")
```

```python
result = []
print(f"numbers: {numbers}\n")

for n in numbers:
    result.append(n * 2)

#using list comprehension
result = [n * 2 for n in numbers]
print(f"result: {result}\n")

#writing a list comprehension to extract only odd numbers from the list
odd_numbers = [n for n in numbers if n % 2 != 0]
print(f"odd numbers: {odd_numbers}\n")


    #using dictionary comprehension
result_dict = {n: n * 2 for n in numbers}
#printing the dictionary
print(f"result dictionary: {result_dict}\n")

#writing a function that accepts any number of marks using *args and returns the average marks
def average_marks(*args):
    total = sum(args)
    count = len(args)
    return total / count


print(f"\nAverage marks: {average_marks(80,0, 90, 100,70, 85)}")


```

```python
#function that accepts student details using kwargs and print them

def print_student_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_student_details(name="Amrit", age=25, course="Data Science", marks=85)
```
