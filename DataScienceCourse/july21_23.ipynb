2.1 .loc - Label Based Selection

selects by label - row index labels and column names

```python
df = pd.DataFrame({
    "name": ["Aayush", "Bina", "Cirag", "Diya"],
    "city": ["Kathmandu", "Pokhara", "Kathmandu", "Biratnagar"],
    "amount": [4500, 8200, 6100, 3000]
}, index=["ord1", "ord2", "ord3", "ord4"])

print(df.loc["ord2"]) # single row by label
print()
print(df.loc["ord2", "amount"])   # single cell by row label + column name
print()
print(df.loc["ord1":"ord3"])           # label-based slice — INCLUSIVE of both ends
```

Note: unlike Python list slicing .loc slices are inclusive on both ends. df.loc["ord1": "ord3"] includes ord3.

2.2 Selecting rows and columns together

```python
print(df.loc["ord1":"ord3", ["name", "amount"]])
print()
print(df.loc[:, "city"])               # all rows, one column
print()
print(df.loc[:, ["name", "city"]])     # all rows, multiple columns
```

2.3) .loc with boolean conditions

```python
# SQL: SELECT * FROM orders WHERE amount > 5000
print(df.loc[df["amount"] > 5000])

# SQL: SELECT name, amount FROM orders WHERE amount > 5000
print(df.loc[df["amount"] > 5000, ["name", "amount"]])
```

 in .loc with boolean conditions, we use row fileter as teh first argument, column selection as the second.

2.4) Setting values with .loc

```python
df.loc[df["amount"] > 8000, "city"] = "Kathmandu (High Value)"
print(df)
```

3) iloc -> Position - Based Selection

-> selects by integer position, always 0-based, regardless of the actual index labels.

```python
print(df.iloc[0])                # first row, by position
print()
print(df.iloc[0, 2])              # row 0, column 2 (position-based, not name)
print()
print(df.iloc[0:2])               # position-based slice — EXCLUSIVE of the end, like normal Python
print()
print(df.iloc[:, 1])              # all rows, column at position 1
print()
print(df.iloc[0:2, [0, 2]])       # rows 0-1, columns at positions 0 and 2
```

use .loc when you know the label(row name or column name) or you're filtering with a condition. Use .iloc when you only know the position.

# 4.Boolean Filtering and Combining Conditions

```python
# Single condition recap
print(df[df["amount"] > 5000])
```

4.2 Multiple conditions ->  & and | , never and/or


Python's native and and or are designed to compare single truth values (true or false), whereas Pandas needs to compare entire lists (Series) of truth values all at once.

Python's native and operator evaluates the truthiness of the entire object as a whole. It looks at the first Series and asks, "Is this entire list true?" Pandas gets confused because a Series has multiple values, resulting in a famous error: ValueError: The truth value of a Series is ambiguous.

```python
# SQL: SELECT * FROM orders WHERE amount > 5000 AND city = 'Kathmandu'
result = df[(df["amount"] > 5000) & (df["city"] == "Kathmandu")]
print(result)

print()
# SQL: SELECT * FROM orders WHERE city = 'Pokhara' OR city = 'Biratnagar'
result2 = df[(df["city"] == "Pokhara") | (df["city"] == "Biratnagar")]
print(result2)
```

4.3 Negation with ~

```python
# SQL: SELECT * FROM orders WHERE NOT city = 'Kathmandu'
print(df[~(df["city"] == "Kathmandu")])
```

5. .query() -> Readable Filtering


```python
print(df.query("amount > 5000"))
print(df.query("amount > 5000 and city == 'Kathmandu'"))
print(df.query("city == 'Pokhara' or city == 'Biratnagar'"))
```

5.2 Refrencing external variables

```python
threshold = 5000
print(df.query("amount > @threshold"))
```

5.3 When to prefer .query() vs boolean masks

-> Use .query() when your filters are simple column comparisons. (e.g., >, <, ==, and, or).

-> If you can write the condition using basic math (>, <, ==) and words (and, or), choose .query().


-> Use boolean masks with .loc when your condition requires custom Python functions or complex logic that cannot be translated into a basic query string.

-> If you have to use .apply(), loops, or a def function you wrote yourself, choose boolean masks.


#

Imagine you want to filter orders where the City name has an even number of vowels. .query() cannot do this because it requires custom Python logic:


1. You must write a custom Python function
def has_even_vowels(text):
    vowels = [char for char in text.lower() if char in 'aeiou']
    return len(vowels) % 2 == 0

2. .query() CANNOT run this easily. You MUST use a boolean mask:
mask = df['City'].apply(has_even_vowels)
even_vowel_orders = df.loc[mask]

# 6. .isin() -> Membership Filtering

6.1 THE SQL IN translation

 this replaces long chains of | conditions — df[(df["city"]=="A") | (df["city"]=="B") | (df["city"]=="C")] becomes one clean .isin(["A","B","C"]) call.

```python
# SQL: SELECT * FROM orders WHERE city IN ('Kathmandu', 'Pokhara')
cities = ["Kathmandu", "Pokhara"]
print(df[df["city"].isin(cities)])
```

6.2 Negated mebership

```python
# SQL: SELECT * FROM orders WHERE city NOT IN ('Kathmandu', 'Pokhara')
print(df[~df["city"].isin(cities)])
```
