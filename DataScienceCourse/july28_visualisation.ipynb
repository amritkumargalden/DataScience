```python
# Histogram

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "age" : [21,20,19,25,24,90,30,31],
    "income": [100000,230000,340000,900000,120000,340000,230000,210000]
})

plt.hist(df["income"], bins = 5, edgecolor = "black")
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR']
data = [23,10,35,15,12]

plt.pie(data, labels = cars, autopct = '%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
# BOX PLOT
data = [
    [10,12,14,15,80,20,22],
    [8,10,11,50,15,19,21],
    [14,16,18,20,26,25,89]
]

plt.boxplot(data)
plt.xlabel("Groups")
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate mock data: 100 random integers centered around 50
data = np.random.normal(50, 10, 100)

# Create the boxplot
plt.boxplot(data, patch_artist=True, boxprops=dict(facecolor='skyblue'))

plt.title('Distribution Analysis Using a Boxplot')
plt.ylabel('Values')
plt.show()
```

```python
df = pd.DataFrame({
    "age": [22, 25, 23, 24, 90, 21, 26, 23],   # one clear outlier
    "income": [30000, 35000, 32000, 31000, 34000, 500000, 33000, 32500],  # another outlier
    "city": ["Kathmandu", "Pokhara", "Kathmandu", "Biratnagar", "Kathmandu", "Pokhara", "Kathmandu", "Pokhara"]
})
print(df.describe())
sns.countplot(x="city", data=df, orient='h')
plt.title("Number of records per city")
plt.show()
```
