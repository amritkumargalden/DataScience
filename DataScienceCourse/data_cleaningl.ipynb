```python
import numpy as np
import pandas as pd
```

```python
messy = pd.DataFrame({
    "name": ["Aayush", "bina", "CIRAG", "Diya ", None],
    "age": ["21", "22", "N/A", "23", "21"],
    "city": ["Kathmandu", "kathmandu", "Pokhara", "Pokhara", "Biratnagar"],
    "amount": [4500, 4500, 8900, np.nan, 3100]
})
print(messy)
```

```python
print(messy.isnull())
print()
print(messy.isnull().sum())
```

```python
print(messy.isnull().mean()*100)
```

```python
# Dropping missing values ->dropna()

df_dropped_rows = messy.dropna()
print(df_dropped_rows)
print(
)

df_dropped_cols = messy.dropna(axis=1)
print(df_dropped_cols)
print()

df_dropped_thresh = messy.dropna(thresh=3)
print(df_dropped_thresh)
print()

df_dropped_subset = messy.dropna(subset=["amount"])
print(df_dropped_subset)
```

```python
df_filled_zero = messy["amount"].fillna(0)
df_filled_mean = messy["amount"].fillna(messy["amount"].mean())
df_filled_median = messy["amount"].fillna(messy["amount"].median())
df_filled_ffill = messy["amount"].ffill()
df_filled_bfill = messy["amount"].bfill()
df_filled_custom = messy["amount"].fillna("Unknown")

print(df_filled_zero)
print()
print(df_filled_mean)
print()
print(df_filled_median)
print()
print(df_filled_ffill)
print()
print(df_filled_bfill)
print()
print(df_filled_custom)
```

```python
messy = pd.DataFrame({
    "name": ["Aayush", "bina", "CIRAG", "Diya Karki ", "Aayush"],
    "age": ["21", "22", "N/A", "23", "21"],
    "city": ["Kathmandu", "Kathmandu", "Pokhara", "Pokhara", "Kathmandu"],
    "amount": [4500, 4500, 8900, np.nan, 4500]
})

print(messy.duplicated())
print()

print(messy.duplicated().sum())
print()
print(messy[messy.duplicated()])
```

```python
print(messy.duplicated(subset=["city"]))
```

```python
df_no_dupes = messy.drop_duplicates()
df_no_dupes_subset = messy.drop_duplicates(subset = ["city"], keep="first")

print()
print(df_no_dupes)

print()
print(df_no_dupes_subset)
```

```python
messy = pd.DataFrame({
    "name": ["Aayush", "bina", "CIRAG ", "  diya karki ", " Aayush"],
    "age": ["21", "22", "N/A", "23", "21"],
    "city": ["Kathmandu", "Kathmandu", "Pokhara", "Pokhara", "Kathmandu"],
    "amount": [4500, 4500, 8900, np.nan, 4500]
})

messy["name_capitalize"] = messy["name"].str.capitalize()

print(messy["name_capitalize"])
print(


)
print(messy)
```

```python
# Case Normalization 

messy["city_clean"] = messy["city"].str.lower()
print(messy["city_clean"])

messy["name_clean"] = messy["name"].str.title()
print(messy["name_clean"])
print()
print(messy)
```

```python
# Whitespace Stripping

messy["name_strip"] = messy["name_capitalize"].str.strip()
print(messy["name_strip"])
```

```python
# Replacing and extracting pattens

messy["city_clean"] = messy["city_clean"].str.replace("pokhara", "Dhading")
print(messy["name"].str.contains("iya", na = False))
print(
    messy["name"].str.len()
)
print(messy["name"].str.split(" ").str[0])

print(messy)
```

```python
messy["name_clean"] = (
    messy["name"]
    .str.strip()
    .str.title()
    .fillna("Unknown")
)

print(messy["name_clean"])
```

```python
# Renaming columns
df_renamed = messy.rename(columns={"amount":"order_amount",
                                   "name":"customer_name"
                                   })

print(df_renamed.columns)
```
