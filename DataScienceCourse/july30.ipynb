```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the built-in iris dataset
iris = sns.load_dataset("iris")

# Set up the figure size for a 3x2 grid
plt.figure(figsize=(12, 10))

# ----------------------------------------------------
# Position 1: Top-Left (Row 1, Column 1)
# ----------------------------------------------------
plt.subplot(3, 2, 1)
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("1. Sepal Length vs Width")

# ----------------------------------------------------
# Position 2: Top-Right (Row 1, Column 2) -> Your target position
# ----------------------------------------------------
plt.subplot(3, 2, 2)
sns.boxplot(data=iris, x="species", y="petal_length", hue="species", legend=False)
plt.title("2. Petal Length by Species")

# ----------------------------------------------------
# Position 3: Bottom-Left (Row 2, Column 1)
# ----------------------------------------------------
plt.subplot(3, 2, 3)
sns.histplot(data=iris, x="petal_width", hue="species", kde=True, multiple="stack")
plt.title("3. Petal Width Distribution")

# ----------------------------------------------------
# Position 4: Bottom-Right (Row 2, Column 2)
# ----------------------------------------------------
plt.subplot(3, 2, 4)
sns.violinplot(data=iris, x="species", y="sepal_length", hue="species", legend=False)
plt.title("4. Sepal Length Violin Plot")
# ----------------------------------------------------
# Position 5: Row 3, Column 1 (Fixed Heatmap)
# ----------------------------------------------------
plt.subplot(3, 2, 5)
# Drop the categorical 'species' column and compute correlation matrix
corr_matrix = iris.drop(columns=["species"]).corr()
sns.heatmap(data=corr_matrix, annot=True, cmap="coolwarm", cbar=False)
plt.title("5. Correlation Heatmap")

# ----------------------------------------------------
# Position 6: Row 3, Column 2 (Fixed Title)
# ----------------------------------------------------
plt.subplot(3, 2, 6)
sns.barplot(data=iris, x="species", y="sepal_length", hue="species", legend=False)
plt.title("6. Mean Sepal Length by Species")


# Automatically adjust spacing between plots so labels don't overlap
plt.tight_layout()

# Display the complete grid
plt.show()
```

```python

import pandas as pd
# Sample DataFrame
data_dict = {
    "Year": [2024, 2025, 2026, 2024, 2025, 2026],
    "Users": [100, 150, 280, 90, 130, 210],
    "Platform": ["iOS", "iOS", "iOS", "Android", "Android", "Android"]
}

df = pd.DataFrame(data_dict)

# Seaborn applies a theme automatically
sns.set_theme(style="darkgrid")
plt.figure(figsize=(10, 5))
# Plot with semantic hue grouping
sns.lineplot(data=df, x="Year", y="Users", hue="Platform", style="Platform", markers=True)

# Display
plt.title('User Growth Comparison')
plt.xticks(rotation=90,ha="right") # horizontal alignment = right()
#this ensures the end of the words aligns perfectly under the x-axis tick marks, instead of crossing over the
plt.show()
```
