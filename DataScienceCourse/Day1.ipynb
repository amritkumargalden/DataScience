```python
class Book:
	def __init__(self, title, author, price):
		self.title = title
		self.author = author
		self.price = price
	def details(self):
		print(f"{self.title} written by {self.author} at the price of {self.price}")
	

#objects of the class Dog
Book1 = Book("Peaky Blinders", "Tommy", 999)
Book1.details()

Book2 = Book("The 14 peaks", "John Doe", 799)
Book2.details()
```

```python
# Inheritance

class Animal: ##Parent class
    def __init__(self, name):
        self.name = name
    def eat(self):
        return f"{self.name} is eating"

## Child class
class Dog(Animal):
    def bark(self):
        return f"{self.name} is barking"

##creating an object of the child class
dog1 = Dog("Ronnie")

print(dog1.eat())  # Inherited method from Animal class
print(dog1.bark())  # Method from Dog class

## Another example of inheritance

class Employee:
    def __init__(self, name, branch):
        self.name = name
        self.branch = branch

class Person(Employee):
    def print_address(self, address):
        return f"{self.name} works in {self.branch} and lives in {address}"

Amrit = Person("Amrit", "Data Science")
Amrit.print_address("New Delhi")
```

```python
class Dog:
    def __init__(self, breed_name):
        self.breed_name = breed_name

    def make_sound(self):
        return "Bark"

class breed(Dog):
    def make_sound(self):
        match self.breed_name:
            case "Labrador":
                return "Bhau bhau"
            case "Beagle":
                return "Woof woof"
            case "Bulldog":
                return "Arf arf"
            case "Golden Retriever":
                return "Ruff ruff"
            case _:
                return "Bark"
            
Tommy = breed("Golden Retriever")
Tommy.make_sound()  # Output: "Ruff ruff"
print(Tommy.make_sound())

Chatur = breed("Beagle")
Chatur.make_sound()  # Output: "Woof woof"
```
