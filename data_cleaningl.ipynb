```python
import numpy as np
import pandas as pd
```

```python
messy = pd.DataFrame({
    "name": ["Aayush", "bina", "CIRAG", "Diya ", None],
    "age": ["21", "22", "N/A", "23", "21"],
    "city": ["Kathmandu", "kathmandu", "Pokhara", "Pokhara", "Biratnagar"],
    "amount": [4500, 4500, 8900, np.nan, 3100]
})
print(messy)
```

```python
print(messy.isnull())
print()
print(messy.isnull().sum())
```

```python
print(messy.isnull().mean()*100)
```

```python
# Dropping missing values ->dropna()

df_dropped_rows = messy.dropna()
print(df_dropped_rows)
print(
)

df_dropped_cols = messy.dropna(axis=1)
print(df_dropped_cols)
print()

df_dropped_thresh = messy.dropna(thresh=3)
print(df_dropped_thresh)
print()

df_dropped_subset = messy.dropna(subset=["amount"])
print(df_dropped_subset)
```

```python
df_filled_zero = messy["amount"].fillna(0)
df_filled_mean = messy["amount"].fillna(messy["amount"].mean())
df_filled_median = messy["amount"].fillna(messy["amount"].median())
df_filled_ffill = messy["amount"].ffill()
df_filled_bfill = messy["amount"].bfill()
df_filled_custom = messy["amount"].fillna("Unknown")

print(df_filled_zero)
print()
print(df_filled_mean)
print()
print(df_filled_median)
print()
print(df_filled_ffill)
print()
print(df_filled_bfill)
print()
print(df_filled_custom)
```
